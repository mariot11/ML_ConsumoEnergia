```python
import numpy as np
import pandas as pd
```

```python
url_gastos = "../data/raw/EPFgastos_2025_decodificado.tab"
url_hogar = "../data/raw/EPFhogar_2025_decodificado.tab"

df_gastos = pd.read_csv(url_gastos, sep="\t")
df_hogar = pd.read_csv(url_hogar, sep="\t")

print(df_gastos.info(show_counts=True))
print(df_hogar.info(show_counts=True, max_cols=None))
```

```python
with pd.option_context("display.max_rows", None, "display.max_columns", None, "display.width", 1000):
    resumen_gastos = pd.DataFrame({
        "dtype": df_gastos.dtypes.astype(str),
        "no_nulos": df_gastos.notna().sum(),
        "nulos_nan": df_gastos.isna().sum(),
        "pct_nulos_nan": (df_gastos.isna().mean() * 100).round(2),
        "n_unicos": df_gastos.nunique(dropna=True),
        "muestra_valores": df_gastos.apply(lambda s: list(s.dropna().unique()[:3]))
    })
    display(resumen_gastos)
```

```python
with pd.option_context("display.max_rows", None, "display.max_columns", None, "display.width", 1000):
    resumen_hogar = pd.DataFrame({
        "dtype": df_hogar.dtypes.astype(str),
        "no_nulos": df_hogar.notna().sum(),
        "nulos_nan": df_hogar.isna().sum(),
        "pct_nulos_nan": (df_hogar.isna().mean() * 100).round(2),
        "n_unicos": df_hogar.nunique(dropna=True),
        "muestra_valores": df_hogar.apply(lambda s: list(s.dropna().unique()[:3]))
    })
    display(resumen_hogar)
```

```python
df_gastos_sin_nulos = df_gastos.dropna(axis= 1, thresh = df_gastos.shape[0]/2)
df_hogar = df_hogar.dropna(axis= 1, thresh= df_hogar.shape[0]/2)

print(df_gastos_sin_nulos.info())
print(df_hogar.info())
```

```python
suministros = [
    "04411", "04511", "04521", "04523",
    "04531", "04541", "04543", "04545", "04548"
]

# Filtrar los suministros
df_gastos_edit = df_gastos_sin_nulos[
    df_gastos_sin_nulos["CODIGO"].isin(suministros)
].copy()

# Calcular el gasto sin elevación espacial
df_gastos_edit["GASTMON_AJUSTADO"] = (
    df_gastos_edit["GASTMON"] / df_gastos_edit["FACTOR"]
)

# Agrupar por hogar
df_gastos_edit = (
    df_gastos_edit
    .groupby("NUMERO", as_index=False)["GASTMON_AJUSTADO"]
    .sum()
)

# Renombrar columnas
df_gastos_edit = df_gastos_edit.rename(
    columns={"GASTMON_AJUSTADO": "GASTO_SUMINISTROS"}
)

# Mostrar resultados
print(df_gastos_edit.head(20))
print(df_gastos_edit["GASTO_SUMINISTROS"].describe())
```

```python
s_todos = pd.Series(df_gastos["NUMERO"]).dropna()
s_todos = s_todos[s_todos != 0]

s_suministros = pd.Series(df_gastos_edit["NUMERO"]).dropna()
s_suministros = s_suministros[s_suministros != 0]

faltantes = s_todos[~s_todos.isin(s_suministros)].unique()

print("Hogares faltantes:", len(faltantes))
print(faltantes)
```
